```python
from pathlib import Path
import pandas as pd

teams_path = Path("/home/giova/workspace/fm26/data/brazil/teams")
raw_html = Path("/home/giova/workspace/fm26/data/brazil/raw/may2026.html")

player_score_dict = {}
for f in teams_path.rglob("*.csv"):
    if f.stem.endswith("first_team"):
        f.rename(f.parent / "team_1.csv")
    if f.stem.endswith("second_team"):
        f.rename(f.parent / "team_2.csv")

max_team_nbr = 0
for f in teams_path.rglob("*.csv"):
    team_nbr = int(str(f.stem).split("_")[-1])
    if team_nbr > max_team_nbr:
        max_team_nbr = team_nbr

print("Max team number:", max_team_nbr)

for f in teams_path.rglob("*.csv"):
    team_nbr = int(str(f.stem).split("_")[-1])
    score = abs(team_nbr - max_team_nbr - 1)
    df = pd.read_csv(f)
    for player in df["Name"]:
        player_score_dict[player] = player_score_dict.get(player, 0) + score

# Parse the raw HTML and get the position and club string for each player
df_raw = pd.read_html(raw_html, encoding="utf-8")[0]
df_raw.columns = df_raw.columns.str.strip()

raw_map = (
    df_raw[["Name", "Position", "Club"]]
    .dropna(subset=["Name"])
    .groupby("Name", as_index=False)
    .first()
    .set_index("Name")
)

player_score_position_club = {
    name: (
        score,
        raw_map.loc[name, "Position"] if name in raw_map.index else None,
        raw_map.loc[name, "Club"] if name in raw_map.index else None,
    )
    for name, score in player_score_dict.items()
}

# Example
for name, (score, position, club) in list(player_score_position_club.items())[:10]:
    print(name, score, position, club)
```

```python
sorted_dict = dict(sorted(player_score_position_club.items(), key=lambda x: x[1][0], reverse=True))
first_5_gks = dict(sorted(
    {k: v for k, v in sorted_dict.items() if v[1] == "GK"}.items(),
    key=lambda x: x[1][0],
    reverse=True
)[:5])
first_55 = dict(list(sorted_dict.items())[:55])
first_23 = dict(list(sorted_dict.items())[:23])
first_55_sorted = dict(sorted(first_55.items(), key=lambda x: x[0].split(" ")[-1], reverse=False))
first_23_sorted = dict(sorted(first_23.items(), key=lambda x: x[0].split(" ")[-1], reverse=False))

non_gk = {k: v for k, v in sorted_dict.items() if v[1] != "GK"}
first_50_non_gk = dict(list(non_gk.items())[:50])
first_20_non_gk = dict(list(non_gk.items())[:20])
first_27_non_gk = dict(list(non_gk.items())[:27])
first_23_non_gk = dict(list(non_gk.items())[:23])

first_50_non_gk_sorted = dict(sorted(first_50_non_gk.items(), key=lambda x: x[0], reverse=False))
first_20_non_gk_sorted = dict(sorted(first_20_non_gk.items(), key=lambda x: x[0], reverse=False))
first_27_non_gk_sorted = dict(sorted(first_27_non_gk.items(), key=lambda x: x[0], reverse=False))
first_23_non_gk_sorted = dict(sorted(first_23_non_gk.items(), key=lambda x: x[0], reverse=False))
```
