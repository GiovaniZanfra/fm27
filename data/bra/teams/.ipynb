```python
from pathlib import Path
import pandas as pd

path = Path("/home/giova/workspace/fm26/data/bra/teams")


player_score_dict = {}
for f in path.rglob("*.csv"):
    if f.stem.endswith("first_team"):
        f.rename(f.parent / f"team_1.csv")
    if f.stem.endswith("second_team"):
        f.rename(f.parent / f"team_2.csv")
    

for f in path.rglob("*.csv"):
    team_nbr = int(str(f.stem).split("_")[-1])
    score = abs(team_nbr - 14 - 1)
    df = pd.read_csv(f)
    for player in df["Name"]:
        if player not in player_score_dict:
            player_score_dict[player] = score
        else:
            player_score_dict[player] += score

player_score_dict["Matheus Thuler"] = 0
player_score_dict["Murilo"] = 0
player_score_dict["Lucas Veríssimo"] = 0
player_score_dict["Luiz Júnior"] = 0
player_score_dict["Talisca"] = 0
player_score_dict["Tetê"] = 0
player_score_dict["Vitão"] = 0
player_score_dict["Sávio"] = 0
player_score_dict["Evanilson"] = 0
player_score_dict["Arthur"] = 0
player_score_dict["Matheus Thuler"] = 0
player_score_dict["Murilo"] = 0
player_score_dict["Lucas Veríssimo"] = 0
player_score_dict["Luiz Júnior"] = 0
player_score_dict["Talisca"] = 0
player_score_dict["Tetê"] = 0
player_score_dict["Vitão"] = 0
player_score_dict["Marcão"] = 0
player_score_dict["David Neres"] = 0
player_score_dict["Malcom"] = 0
```
